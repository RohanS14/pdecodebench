```python
import pandas as pd
```

```python
df = pd.read_excel("data/physics_code.xlsx")
```

```python
df.shape
```

```python
df.head(5)
```

```python
if "obf_vars" in df.columns:
    del df["obf_vars"]
if "obf_comments" in df.columns:
    del df["obf_comments"]
if "obf_comments" in df.columns:
    del df["obf_comments"]
```

### Fix columns

```python
# Add gt_sample column
df["gt_sample"] = df["title"].str.split("_").str[0] + "_" + df["title"].str.split("_").str[-1]
df["gt_sample"] = df["gt_sample"].astype(str)
```

```python
df["gt_sample"].value_counts()
```

```python
df["phys_valid"].value_counts()
```

```python
# Change to boolean
if type(df["phys_valid"][0]) == str:
    df["phys_valid"] = (df["phys_valid"] == "Yes")
df["phys_valid"].value_counts()
```

```python
# Add mod_type column
df["mod_type"] = df["title"].str.split("_").str[1] + "_" + df["title"].str.split("_").str[2]
df["mod_type"] = df["mod_type"].astype(str)
df["mod_type"].value_counts()
```

```python
df.dtypes
```

```python
df["pde_class_1"] = df["pde_class"].str.replace(r"\b([A-Za-z]+) Equation\b", r"\1", regex=True)
df["pde_class_1"] = df["pde_class_1"].str.lower().str.replace(" ", "-", regex=False)
df["pde_class"] = df["pde_class_1"]
del df["pde_class_1"]
df["pde_class"].value_counts()
```

```python
df["phys_process"].value_counts()
```

```python
# Replace - with /
df["phys_process"] = df["phys_process"].str.lower().str.replace("-", "/", regex=False)
df["phys_process"].value_counts()
```

```python
# Replace - with /
df["num_method"] = df["num_method"].str.lower().str.replace("-", "/", regex=False)
df["num_method"].value_counts()
```
